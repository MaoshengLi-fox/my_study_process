## Task 1 (2 points)
Implement a function to draw a graph of the Empirical Cummulative Distribution Function. Illustrate its use by taking a random smaple from the standard normal distribution $\mathcal{N}(0, 1)$ and drawing a plot of both true and empirical CDFs.

```python
# TODO: Your code here
def plot_empirical_cdf(sample, true_cdf_func, label_emp="Empirical CDF", label_true="True CDF"):

    sorted_sample = np.sort(sample)

    ecdf = np.arange(1, len(sorted_sample) + 1) / len(sorted_sample)

    plt.step(sorted_sample, ecdf, where='post', label=label_emp)
    x = np.linspace(min(sorted_sample) - 1, max(sorted_sample) + 1, 300)
    plt.plot(x, true_cdf_func(x), 'r--', lw=2, label=label_true)

    plt.title("Empirical vs True Cumulative Distribution Function")
    plt.xlabel("x")
    plt.ylabel("CDF")
    plt.legend()
    plt.grid(True)
    plt.show()

sample = np.random.normal(0, 1, 1000)
plot_empirical_cdf(sample, scipy.stats.norm.cdf)
```

## Task 2 (2 point)
Review the DKW Inequality introduce in the lectures. Using that result and assuming that size of the sample $n$ and miscoverage parameter $\alpha$ are known, find the value $\epsilon$ such that:
$$\mathbb{P}\left\{||F_n - F||_{\infty} \gt \epsilon \right\} \le \alpha$$

Your solution:
  - Given $n$ and $\alpha$, find $\varepsilon$ such that  
  $$
  P\left( \| F_n - F \|_\infty > \varepsilon \right) \le \alpha
  $$
  - According to the DKW inequality  
  $$
  P\left( \sqrt{n} \| F_n - F \|_\infty > z \right) \le 2 e^{-2z^2}
  $$  
  - Let $z = \sqrt{n} \varepsilon$. Then,  
  $$
  P\left( \| F_n - F \|_\infty > \varepsilon \right) = P\left( \sqrt{n} \| F_n - F \|_\infty > \sqrt{n} \varepsilon \right) \le 2 e^{-2n\varepsilon^2}
  $$
  - So we want  $2 e^{-2n\varepsilon^2} \le \alpha$
    - $$e^{-2n\varepsilon^2} \le \frac{\alpha}{2}$$
    - $$-2n\varepsilon^2 \le \ln \frac{\alpha}{2}$$
    - $$\varepsilon^2 \ge -\frac{1}{2n} \ln \frac{\alpha}{2}$$
  - In the end  
  $$
  \varepsilon = \sqrt{ \frac{1}{2n} \ln \frac{2}{\alpha} }
  $$

In the previous task you have found the width of a fixed-length *confidence interval* for the CDF. Illustrate this result using your solution for Task 1. Consider a Laplace distribution with parameters $\mu=0$ and $b=2$, a sample of size $n=35$ and miscoverage level $\alpha=0.05$. Plot true and empirical CDFs with a confidence interval around the empirical CDF, where the width of the interval is given by your solution to Task 2.

```python
# TODO: Your code here
mu, b = 0.0, 2.0
n = 35
alpha = 0.05

sample = np.random.laplace(loc=mu, scale=b, size=n)
x_sorted = np.sort(sample)
ecdf = np.arange(1, n + 1) / n

eps = np.sqrt(np.log(2/alpha) / (2*n))

x_plot = np.linspace(x_sorted.min() - 4*b, x_sorted.max() + 4*b, 800)
true_cdf = scipy.stats.laplace.cdf(x_plot, loc=mu, scale=b)

x_step = np.r_[x_sorted[0]-1e-9, x_sorted]
y_step = np.r_[0, ecdf]

lower = np.clip(y_step - eps, 0, 1)
upper = np.clip(y_step + eps, 0, 1)

plt.figure(figsize=(7,5))

plt.plot(x_plot, true_cdf, lw=2, label='True CDF')

plt.step(x_step, y_step, where='post', label='Empirical CDF', linewidth=1.8)

plt.fill_between(x_step, lower, upper, where=upper>=lower, alpha=0.25,
                 step='post', label=f'Fixed-width CI')

plt.title(f'ECDF with Fixed-width CI via DKW (n={n}, α={alpha})')
plt.xlabel('x'); plt.ylabel('CDF')
plt.ylim(-0.02, 1.02)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Task 1 (3 points)
Find the KL divergence between two Laplace distributions with parameters $(\mu_1, b_1)$ and $(\mu_2, b_2)$. 

- make 
$$
p(x) = \frac{1}{2b_1} e^{-\frac{|x-\mu_1|}{b_1}} \\
q(x) = \frac{1}{2b_2} e^{-\frac{|x-\mu_2|}{b_2}}
$$
- then
$$
\begin{aligned}
D_{KL}(p\|q)&=\int p(x) \log \frac{p(x)}{q(x)} \\
            &=\log\frac{b_2}{b_1}+ \frac{1}{b_2}\mathbb{E}_p|X-\mu_2| - \frac{1}{b_1}\mathbb{E}_p|X-\mu_1|
\end{aligned}
$$
- because
$$
\begin{aligned}
\mathbb{E}_p|X-\mu_1| &=b_1 \\
\mathbb{E}_p|X-\mu_2| &=\mathbb{E}_p|X-\mu_1+\mu_1-\mu_2| \\
                      &=\mathbb{E}_p|Z-d| \quad  Z=(X-\mu_1) \sim Lap(0,b_1),\quad d=\mu_1-\mu_2 \\
                      &=\frac{1}{2b_1}[\int_{-\infty}^{0} (d-z)e^{z/b_1}dz+\int_{0}^{d} (d-z)e^{-z/b_1}dz+\int_{d}^{\infty} (d-z)e^{-z/b_1}dz] \\
                      &=d+b_1 e^{-d/b_1}
\end{aligned}
$$

- in the end
$$
D_{KL}(p\|q)=\log\!\frac{b_2}{b_1} + \frac{|\mu_1-\mu_2|}{b_2} + \frac{b_1}{b_2} e^{-|\mu_1-\mu_2|/b_1}- 1
$$

## Taks 2* (2 bonus points)
Implement the calculation of KL divergence between two 1d distributions numericaly (assuming that you have access to density). Using a grid to approximate the corresponding integral. Illustrate with the case from the previous task: two Laplace sitributions.

```python
# TODO: Your coded here
def kl_divergence_grid(pdf_p, pdf_q, x_min, x_max, num=200_000, eps=1e-300):
    x = np.linspace(x_min, x_max, num, dtype=float)
    p = np.clip(pdf_p(x), 0.0, None) + eps
    q = np.clip(pdf_q(x), 0.0, None) + eps
    integrand = p * (np.log(p) - np.log(q))
    return np.trapezoid(integrand, x)

mu1, b1 = 0.0, 1.0
mu2, b2 = 1.0, 1.5

pdf_p = lambda x: scipy.stats.laplace.pdf(x, loc=mu1, scale=b1)
pdf_q = lambda x: scipy.stats.laplace.pdf(x, loc=mu2, scale=b2)

L = 13 * max(b1, b2)
left = min(mu1, mu2) - L
right = max(mu1, mu2) + L

kl_num = kl_divergence_grid(pdf_p, pdf_q, left, right, num=300_000)
print(f"KL(P||Q) = {kl_num:.6f}")
```

# Part 3: Maximum likelihood estimation (3 points)
## Task 1 (1.5 points)
Find the maximum likelihood estimate of the parameter $\lambda$ of the Exponential distribution.

Your solution:
 - exponential distribution: $f(x | \lambda) = \lambda e^{- \lambda x}$ ,$x \ge 0$,$\lambda >0$
 - likelihood estimation
 $$
 L(\lambda) = \prod_{i=1}^n \lambda e^{-\lambda x_i} = \lambda^n e^{-\lambda \sum_{i=1}^n x_i}
 $$
 - Log-likelihood estimation
 $$
 \ell(\lambda) = \log L(\lambda) = n \log \lambda - \lambda \sum_{i=1}^n x_i
 $$
 - take the derivative and set it to zero
 $$
 \frac{d\ell}{d\lambda} = \frac{n}{\lambda} - \sum_{i=1}^n x_i = 0 \\
 \widehat{\lambda} = \frac{n}{\sum_{i=1}^n x_i}= \frac{1}{\bar{x}}
 $$
 - in the end $\ell''(\lambda) = -\frac{n}{\lambda^2} < 0$
 $$
 \widehat{\lambda}= \frac{1}{\bar{x}}
 $$

## Task 2 (1.5 points)
Find the maximum likelihood estimate of the parameter $\lambda$ of the Poisson distribution.

Your solution:
 - poisson distribution: $f(x | \lambda) = e^{- \lambda} \frac{\lambda^x}{x!}$ ,$x \in \{0,1,2...\}$,$\lambda >0$
 - likelihood estimation
 $$
 L(\lambda) = \prod_{i=1}^n e^{-\lambda} \frac{\lambda^{x_i}}{x_i!} = e^{-n\lambda} \frac{\lambda^{\sum_{i=1}^n x_i}}{\prod_{i=1}^n x_i!}
 $$
 - Log-likelihood estimation
 $$
 \ell(\lambda) = \log L(\lambda) = -n\lambda + \Big(\sum_{i=1}^n x_i\Big)\log \lambda - \sum_{i=1}^n \log(x_i!)
 $$
 - take the derivative and set it to zero
 $$
 \frac{d\ell}{d\lambda} = -n + \frac{\sum_{i=1}^n x_i}{\lambda} = 0 \\
 \widehat{\lambda} = \frac{1}{n} \sum_{i=1}^n x_i = \bar{x}
 $$
 - in the end $\ell''(\lambda) = -\frac{\sum_{i=1}^n x_i}{\lambda^2} < 0$
 $$
 \widehat{\lambda}_{\text{Pois}} = \bar{x}
 $$
