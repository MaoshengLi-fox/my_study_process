##### (b)

```python
import numpy as np

np.random.seed(42)
n = 10000
# prevalence
disease = np.zeros(n, dtype=bool)
for i in range(n):
    if np.random.rand() < 0.01:
        disease[i] = True

# test
test_positive = np.zeros(n, dtype=bool)
for i in range(n):
    if disease[i]:
        if np.random.rand() < 0.95:
            test_positive[i] = True
    else:
        if np.random.rand() < 0.02:
            test_positive[i] = True

# P(D|T)
if test_positive.sum() > 0:
    p_hat = disease[test_positive].sum() / test_positive.sum()
else:
    p_hat = float('nan')

print(f"P(D|T) = {p_hat:.4f}")
```

##### (C)

```python
import matplotlib.pyplot as plt

np.random.seed(42)
n_reps = 1000
n=100
estimates = []

for r in range(n_reps):
    disease = np.zeros(n, dtype=bool)
    for i in range(n):
        if np.random.rand() < 0.01:
            disease[i] = True

    test_positive = np.zeros(n, dtype=bool)
    for i in range(n):
        if disease[i]:
            if np.random.rand() < 0.95:
                test_positive[i] = True
        else:
            if np.random.rand() < 0.02:
                test_positive[i] = True

    pos = int(test_positive.sum())
    if pos > 0:
        estimates.append(disease[test_positive].sum() / pos)

plt.hist(estimates, bins=30, edgecolor="black")
plt.xlabel("Estimated P(D|T)")
plt.ylabel("Frequency")
plt.title("histogram of P(D|T) estimates (1000 repetitions)")
plt.show()
```
